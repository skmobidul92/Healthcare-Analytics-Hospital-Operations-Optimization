```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("healthcare_analytics.csv")
df.head()
print(df.shape)
df.info()
df.describe()
print(df.columns)

df.isnull().sum()
df.duplicated().sum()
df = df.drop_duplicates()

df["Admission_Date"] = pd.to_datetime(df["Admission_Date"])
df["Discharge_Date"] = pd.to_datetime(df["Discharge_Date"])
df["Appointment_Date"] = pd.to_datetime(df["Appointment_Date"])
df.info()

df["Admission_Month"] = df["Admission_Date"].dt.month_name()
df["Admission_Day"] = df["Admission_Date"].dt.day_name()
df["Appointment_Day"] = df["Appointment_Date"].dt.day_name()
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 35, 50, 65, 100],
    labels=["0-18", "19-35", "36-50", "51-65", "65+"])
df.head()
```

```python
#Hospital Readmission Prediction
#Readmission Distribution

df["Readmission"].value_counts()
readmission_rate = (df["Readmission"].value_counts(normalize=True) * 100)
print(readmission_rate)

sns.countplot(data=df,x="Readmission")
plt.title("Patient Readmission Distribution")
plt.xlabel("Readmission")
plt.ylabel("Number of Patients")
plt.show()

#Readmission by Department
department_readmission = pd.crosstab(df["Department"],df["Readmission"],normalize="index") * 100
print(department_readmission)

department_readmission.plot(kind="bar")
plt.title("Readmission Rate by Department")
plt.xlabel("Department")
plt.ylabel("Percentage")
plt.xticks(rotation=45)
plt.show()

#Readmission vs Length of Stay

sns.boxplot(data=df,x="Readmission",y="Length_of_Stay_Days")
plt.title("Length of Stay vs Readmission")
plt.show()
```

```python
#OPD Analytics Dashboard
#Appointment Status

df["Appointment_Status"].value_counts()
sns.countplot(data=df,x="Appointment_Status")
plt.title("Appointment Status Distribution")
plt.xticks(rotation=45)
plt.show()

#Patients by Department

department_count = df["Department"].value_counts()
print(department_count)
sns.countplot(data=df,y="Department",order=df["Department"].value_counts().index)
plt.title("Number of Patients by Department")
plt.xlabel("Number of Patients")
plt.ylabel("Department")

#Average Waiting Time by Department

avg_waiting_time = (df.groupby("Department")["Waiting_Time_Minutes"].mean().sort_values(ascending=False))
print(avg_waiting_time)

avg_waiting_time.plot(kind="bar")
plt.title("Average Waiting Time by Department")
plt.xlabel("Department")
plt.ylabel("Average Waiting Time (Minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
#Appointment Scheduling Optimization
#Appointment Count by Day

appointment_day = (df["Appointment_Day"].value_counts())
print(appointment_day)
days_order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
appointment_day = (df["Appointment_Day"].value_counts().reindex(days_order))

appointment_day.plot(kind="bar")
plt.title("Appointments by Day")
plt.xlabel("Day")
plt.ylabel("Number of Appointments")
plt.xticks(rotation=45)
plt.show()

#Average Waiting Time by Appointment Day

day_waiting_time = (df.groupby("Appointment_Day")["Waiting_Time_Minutes"].mean().reindex(days_order))
print(day_waiting_time)

day_waiting_time.plot(kind="line",marker="o",)
plt.title("Average Waiting Time by Appointment Day")
plt.xlabel("Day")
plt.ylabel("Waiting Time (Minutes)")
plt.grid()
plt.show()
```

```python
#Pharmacy Demand Forecasting
#Medicine Demand Analysis

medicine_demand =(df.groupby("Pharmacy_Medicine")["Medicine_Demand_Units"].sum().sort_values(ascending=False))
print(medicine_demand)

medicine_demand.plot(kind="bar")
plt.title("Total Pharmacy Medicine Demand")
plt.xlabel("Medicine")
plt.ylabel("Total Demand Units")
plt.xticks(rotation=45)
plt.show()

#Average Medicine Demand

average_medicine_demand = (df.groupby("Pharmacy_Medicine")["Medicine_Demand_Units"].mean().sort_values(ascending=False))
print(average_medicine_demand)

sns.barplot(x=average_medicine_demand.index,y=average_medicine_demand.values)
plt.title("Average Medicine Demand by Medicine")
plt.xlabel("Medicine")
plt.ylabel("Average Demand")
plt.xticks(rotation=45)
plt.show()

#Pharmacy Demand Over Time

df["Admission_Month_Number"] = df["Admission_Date"].dt.month
monthly_demand = (df.groupby("Admission_Month_Number")["Medicine_Demand_Units"].sum())
print(monthly_demand)


plt.plot(monthly_demand.index,monthly_demand.values,marker="o")
plt.title("Monthly Pharmacy Demand")
plt.xlabel("Month")
plt.ylabel("Medicine Demand Units")
plt.xticks()
plt.grid()
plt.show()
```

```python
#ICU Bed Optimization
#ICU Patient Count

df["ICU_Required"].value_counts()

sns.countplot(data=df,x="ICU_Required")
plt.title("ICU Requirement Distribution")
plt.show()

#Average Bed Occupancy by Department
bed_occupancy = (df.groupby("Department")["Bed_Occupancy_Percent"].mean().sort_values(ascending=False))
print(bed_occupancy)

bed_occupancy.plot(kind="bar")
plt.title("Average Bed Occupancy by Department")
plt.xlabel("Department")
plt.ylabel("Bed Occupancy Percentage")
plt.xticks(rotation=45)
plt.show()

#ICU Requirement vs Length of Stay

sns.boxplot(data=df,x="ICU_Required",y="Length_of_Stay_Days")
plt.title("ICU Requirement vs Length of Stay")
plt.show()
```
